## **Import**

```python
import pandas as pd
import upstox_client
```

```python
access_token = ""
```

```python
configuration = upstox_client.Configuration()
configuration.access_token = access_token
apiInstance = upstox_client.HistoryV3Api(upstox_client.ApiClient(configuration))
```

```python
try:
    response = apiInstance.get_historical_candle_data1("NSE_EQ|INE002A01018", "minutes", "1", "2022-02-01", "2022-01-01")
    print("Data fetched successfully, total rows fetched:", len(response.data.candles))
except Exception as e:
    print("Exception when calling HistoryV3Api->get_historical_candle_data1: %s\n" % e)
```

```python
# Convert the response data to a pandas DataFrame
columns = ["timestamp", "open", "high", "low", "close", "volume", "open_interest"]

df = pd.DataFrame(response.data.candles, columns=columns)
df["timestamp"] = pd.to_datetime(df["timestamp"])
df = df.sort_values("timestamp", ascending=True).reset_index(drop=True)

df.head()
```

```python
# Get all trading dates in the dataframe
trading_dates = df["timestamp"].dt.date.drop_duplicates().reset_index(drop=True)

trading_dates
```
